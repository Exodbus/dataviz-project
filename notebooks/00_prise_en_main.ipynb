Ce notebook montre comment utiliser le socle pour explorer les données avant de construire une page d'axe.

Prérequis : `uv run python scripts/download.py` a été lancé une fois.

## 1. Charger une année harmonisée

```python
from so_ece.dataset import load_year

d25, d26 = load_year(2025), load_year(2026)
d25.shape, d26.shape
```

## 2. Choisir une population

Les mêmes filtres que dans l'app : région, rôles data/IA, expérience.

```python
from so_ece.population import Filters, select_data_profiles, select_other_devs

f = Filters(region="Europe")
data25, data26 = select_data_profiles(d25, f), select_data_profiles(d26, f)
autres26 = select_other_devs(d26, f)
len(data25), len(data26), len(autres26)
```

## 3. Une part, avec son effectif

Le dénominateur est toujours le nombre de répondants à la question, pas le nombre total de répondants.

```python
from so_ece.analysis.multi import share, shares_table

share(data26, "LanguagesHaveWorkedWith", "Python")
```

```python
shares_table(data26, "DatabasesHaveWorkedWith").head(10)
```

## 4. Une évolution 2025 → 2026, avec son intervalle à 95 %

```python
from so_ece.analysis.compare import compare_items

compare_items(data25, data26, "DatabasesHaveWorkedWith", ["DuckDB", "PostgreSQL", "MySQL"]).round(1)
```

## 5. Le même calcul chez les autres développeurs (groupe de contrôle)

```python
autres25 = select_other_devs(d25, f)
compare_items(autres25, autres26, "DatabasesHaveWorkedWith", ["DuckDB", "PostgreSQL", "MySQL"]).round(1)
```
